# Classification: The Perceptron
We will now study the problem of **classification**, i.e. assigning a data point $\mathbf{x}$ to one of $K$ classes $\mathcal{C}_k$, $k = 1, \dots, K$ (for simplicity, we will assume the classes to be disjoint).

The simplest approach to classification is to use a **discriminant function**, i.e. a function that directly assigns the input $\mathbf{x}$ to one of $K$ classes $\mathcal{C}_k$. For simplicity, from now on we will only consider the binary cases, $K = 2$.
The model you are going to implement in this exercise is the **perceptron**, which models a single artificial neuron as follows:
* The neuron has some **weights** $\mathbf{w}$ and a bias $w_0$, representing the threshold needed to activate the neuron
* An input $\mathbf{x}$ gets modulated by the neuron weights through a **weighted sum** $\mathbf{w}^\intercal \mathbf{x}$, which is then compared against the bias
* If $\mathbf{w}^\intercal \mathbf{x} + w_0 \ge 0$, then the neuron "fires" (i.e. is activated), otherwise it does not

More formally, a perceptron represents the following linear discriminant function:
$$
  y(\mathbf{x}) = \operatorname{sgn}(\phi\bigl(\mathbf{x})\bigr) = \begin{cases}
              +1 &\phi(\mathbf{x}) \ge 0, \\
              -1 &\phi(\mathbf{x}) < 0,
          \end{cases}
$$
where we introduced the **neuron potential** $\phi(\mathbf{x}) \equiv \mathbf{w}^\intercal\mathbf{x} + w_0$.

```
import numpy as np
import plotly.graph_objects as go


def generate_two_gaussians(n, means=[[0, 0], [1, 1]], cov0=[[0.05, 0.01], [0.01, 0.05]], cov1=[[0.03, 0.02], [0.02, 0.03]], rng=np.random.default_rng(42)):
  X0 = rng.multivariate_normal(mean=means[0], cov=cov0, size=n)
  X1 = rng.multivariate_normal(mean=means[1], cov=cov1, size=n)
  X = np.vstack([X0, X1])
  y = np.hstack([-np.ones(n), np.ones(n)])
  return X, y

X, y = generate_two_gaussians(100)

plot_two_gaussians(X, y)
```

## The Perceptron Learning Algorithm
Given a linearly separable dataset of input-observation pairs, $\mathcal{D} = \{(\mathbf{x}^{(1)}, y^{(1)}), \dots, (\mathbf{x}^{(n)}, y^{(n)}) \}$, we can identify a learning rule, the Perceptron Learning Algorithm (PLA), which finds a set of weights $\mathbf{w}$ correctly classifying all data points. You have seen that the update rule for the weights is
  $$\mathbf{w} \leftarrow \mathbf{w} + y^* \mathbf{x}^*,$$
where $\mathbf{x}^*$ is a misclassified example and $y^*$ the corresponding true label (we will see more about this in class).

Recall that the update rule can be derived as an application of **Stochastic Gradient Descent** (SGD) to an appropriate loss function. To derive it, note that all correctly classified examples satisfy
$$y \,\mathbf{w}^\intercal \mathbf{x} > 0,$$
so, for an intuitive function $\mathcal{L}(\mathbf{w})$ to be minimized, we require:
* For correctly classified examples, $\mathcal{L}(\mathbf{w})$ should be zero
* For misclassified examples, we want to minimize $-y\,\mathbf{w}^\intercal\mathbf{x}$

Which leads us the the **perceptron loss**
$$\mathcal{L}(\mathbf{w}) = -\sum_{i \in \mathcal{M}}y^{(i)} \mathbf{w}^\intercal\mathbf{x}^{(i)}$$
where $\mathcal{M}$ is the set of all misclassified examples. Applying the SGD algorithm to this loss function leads precisely to the perceptron update rule.

## Logistic Regression
As you have seen in class, the standard Perceptron presents a certain number of issues: in particular, it never converges when data is not linearly separable due to the hard thresholding, and there is no probabilistic interpretation of its output.

This can be solved by allowing a smooth activation instead of using a step function: in particular, the sigmoid function
 $$f(x) = \frac{e^x}{1+e^x}$$
squashes the output between 0 and 1, and being differentiable we can directly apply GD to the negative log-likelihood. The weight update rule, in this case, is
  $$\mathbf{w} \leftarrow \mathbf{w} + \eta \nabla_\mathbf{w} \ell(\mathbf{w}),$$
where $\eta$ is the learning rate and $\ell(\mathbf{w})$ is the log-likelihood:
  $$\ell(\mathbf{w}) = \sum_{i=1}^m[c_i \log \pi_i + (1 - \pi_i)\log(1 - \pi_i)].$$

#### Intermezzo: Feature Scaling
A common problem when working with real data is that often different features have very different scales, which is not desirable with most machine learning algorithms. You can immediately see it in the dataset description: the mean area goes from 143.5 to 2501.0, while the mean smoothness from 0.053 to 0.163. Without a proper rescaling, larger features will probably dominate, slowing down or preventing convergence.

The most common feature scaling methods are **normalization** and **standardization**:
* In normalization (or min-max scaling), features are transformed as follows:
  $$x' = \frac{x - x_\mathrm{min}}{x_\mathrm{max} - x_\mathrm{min}},$$
so that the new features range from 0 to 1. Scikit-Learn provides a `MinMaxScaler` transformer to implement this.
* In standardization, features are transformed by subtracting the mean and dividing by the standard deviation:
  $$x' = \frac{x - \mu}{\sigma},$$
so that the new features have zero mean and unit standard deviation. Scikit-Learn provides a `StandardScaler` transformer to implement this.

**Important**: recall to use the `fit` methods only on the training data. You should first train your transformer on the training set, and only afterwards apply it to the test set!
